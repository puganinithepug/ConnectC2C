```python
import pandas as pd 

from faker import Faker

fake = Faker()

people = [
    {
        "first_name": fake.first_name(),
        "last_name": fake.last_name(),
        "email": fake.unique.email(),
    }
    for _ in range(50)
]

df = pd.DataFrame(people)
df.to_csv("people.csv", index=False)
```
